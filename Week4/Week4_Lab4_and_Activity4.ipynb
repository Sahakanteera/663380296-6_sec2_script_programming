# lab4_1.py - Student Management System (List Operations)

```python
# lab4_1.py - Student Management System (List Operations)
# หมายเหตุ: โค้ดนี้ตัดส่วน "Challenge" (ระบบเมนูแบบ while loop) ออก
#          เพื่อให้นักศึกษานำ concept ด้านล่างไปต่อยอดคิดเองเป็นแบบฝึกหัด

print("--- Lab 4.1: Student Management System ---")

# 1) เริ่มต้นด้วย list ว่างสำหรับเก็บชื่อนักศึกษา
student_names = []

# 2) เพิ่มนักศึกษา: รับชื่อจากผู้ใช้ 3 คน แล้ว append() เข้า list
print("\n[1] Add Students")
for i in range(3): #กำหนดตัวแปรใน loop 👈ต้องทำ
    name = input(f"Enter student name #{i + 1}: ") #กำหนดตัวแปร 👈ต้องทำ
    student_names.append(name)

print(f"Current list: {student_names}")

# 3) ค้นหานักศึกษา: ใช้ 'in' เพื่อเช็คว่ามีชื่อนี้อยู่ใน list หรือไม่
#    ถ้าพบ ให้แสดง index ด้วย index()
print("\n[2] Find a Student")
search_name = input("Enter a name to search for: ")

if search_name in student_names: #เขียนเงื่อนไข ถ้าหากมีชื่อในลิส: 👈ต้องทำ
    position = student_names.index(search_name)
    print(f"Found '{search_name}' at index {position}")
else:
    print(f"'{search_name}' not found in the list")

# 4) ลบนักศึกษา: เช็คก่อนว่ามีอยู่จริงหรือไม่ (กัน error) แล้วค่อย remove()
print("\n[3] Remove a Student")
remove_name = input("Enter a name to remove: ")

if remove_name in student_names:
    student_names.remove(remove_name) #เติมฟังก์ชัน remove(remove_name) 👈ต้องทำ
    print(f"'{remove_name}' has been removed")
else:
    print(f"'{remove_name}' not found, nothing removed")

print(f"Updated list: {student_names}")

# 5) เรียงลำดับรายชื่อตามตัวอักษรด้วย sort()
print("\n[4] Sort Students")
student_names.sort()
print(f"Sorted list: {student_names}")

# 6) นับจำนวนนักศึกษาทั้งหมดด้วย len()
print("\n[5] Count Students")
print(f"Total students: {len(student_names)}") #ใช้คำสั่งlen() หาจำนวนนักศึกษาในลิส student_names👈ต้องทำ

# --------------------------------------------------------------------
# ให้นักศึกษาลองต่อยอดเอง (Challenge):
# ลองเขียนระบบเมนูด้วย while loop (จาก Week 3) ที่ให้ผู้ใช้เลือกได้ว่า
# จะเพิ่ม / ลบ / ค้นหา / เรียงลำดับ / หรือออกจากโปรแกรม
# โดยนำโค้ดแต่ละส่วนด้านบนไปจัดเป็นฟังก์ชันหรือเงื่อนไขในเมนู
# --------------------------------------------------------------------
```

## **Challenge: Menu-driven Student Management System**

This section implements the challenge: a menu-driven system using a `while` loop, allowing interactive management of student names.

```python
student_names = [] # Re-initialize for the menu system, or use the last state of student_names from previous cell

while True:
    print("\n--- Student Management Menu ---")
    print("1. Add Students")
    print("2. Find a Student")
    print("3. Remove a Student")
    print("4. Sort Students")
    print("5. Count Students")
    print("6. Exit")

    choice = input("Enter your choice (1-6): ")

    if choice == '1':
        while True:
            try:
                num_students_to_add = int(input("How many students do you want to add? "))
                if num_students_to_add < 0:
                    print("Please enter a non-negative number.")
                else:
                    break
            except ValueError:
                print("Invalid input. Please enter a number.")
        for i in range(num_students_to_add):
            name = input(f"Enter student name #{len(student_names) + 1}: ")
            student_names.append(name)
        print(f"Current list: {student_names}")
    elif choice == '2':
        search_name = input("Enter a name to search for: ")
        if search_name in student_names:
            position = student_names.index(search_name)
            print(f"Found '{search_name}' at index {position}")
        else:
            print(f"'{search_name}' not found in the list")
    elif choice == '3':
        remove_name = input("Enter a name to remove: ")
        if remove_name in student_names:
            student_names.remove(remove_name)
            print(f"'{remove_name}' has been removed")
        else:
            print(f"'{remove_name}' not found, nothing removed")
        print(f"Updated list: {student_names}")
    elif choice == '4':
        student_names.sort()
        print(f"Sorted list: {student_names}")
    elif choice == '5':
        print(f"Total students: {len(student_names)}")
    elif choice == '6':
        print("Exiting Student Management System. Goodbye!")
        break
    else:
        print("Invalid choice. Please enter a number between 1 and 6.")
```

# lab4_2.py - Geometric Calculations (Tuples & Immutability)

```python
# lab4_2.py - Geometric Calculations (Tuples & Immutability)
# หมายเหตุ: โค้ดนี้ตัดส่วน "Challenge" (รับพิกัดจากผู้ใช้เพื่อคำนวณระยะทาง) ออก
#          เพื่อให้นักศึกษานำ concept ด้านล่างไปต่อยอดคิดเองเป็นแบบฝึกหัด

import math

print("--- Lab 4.2: Geometric Calculations ---")

# 1) กำหนดจุดสองจุดด้วย tuple: point1 และ point2
point1 = (1, 2) #เติมพิกัด x y ของจุด 👈ต้องทำ
point2 = (4, 6)  #เติมพิกัด x y ของจุด 👈ต้องทำ

# 2) เข้าถึงค่าพิกัด x, y ของแต่ละจุด
print("\n[1] Access Coordinates")
print(f"Point 1: x = {point1[0]}, y = {point1[1]}") #เติมเลข 0 และ 1 ใน [] 👈ต้องทำ
print(f"Point 2: x = {point2[0]}, y = {point2[1]}") #เติมเลข 0 และ 1 ใน [] 👈ต้องทำ

# 3) ทดลองแก้ไขค่าใน tuple เพื่อแสดงให้เห็นว่า tuple เปลี่ยนแปลงไม่ได้ (immutable)
print("\n[2] Attempt Modification (Immutability Demonstration)")
try:
    point1[0] = 5  # บรรทัดนี้จะทำให้เกิด TypeError เพราะ tuple แก้ไขไม่ได้
except TypeError as e:
    print(f"Error trying to modify tuple: {e}")

# 4) คำนวณระยะห่างระหว่างจุดสองจุดด้วยสูตรระยะทาง
#    distance = sqrt( (x2 - x1)^2 + (y2 - y1)^2 )
print("\n[3] Calculate Distance")
distance = math.sqrt((point2[0] - point1[0]) ** 2 + (point2[1] - point1[1]) ** 2) #เติมเลขตำแหน่งในลิส 0 และ 1 ใน [] ให้ถูกต้อง 👈ต้องทำ
print(f"Distance between point1 {point1} and point2 {point2}: {distance}")

# --------------------------------------------------------------------
# ให้นักศึกษาลองต่อยอดเอง (Challenge):
# ลองแก้โค้ดให้รับพิกัดของจุดใหม่ 2 จุดจากผู้ใช้ผ่าน input()
# (อย่าลืมแปลงชนิดข้อมูลจาก input() ให้เป็นตัวเลขก่อนนำไปคำนวณ)
# แล้วคำนวณระยะห่างระหว่างจุดทั้งสองที่ผู้ใช้กรอกเข้ามา
# --------------------------------------------------------------------
```

## **Challenge: Calculate Distance from User Input**

This section addresses the challenge of `lab4_2.py` by prompting the user for two points' coordinates and calculating the Euclidean distance between them.

```python
import math

print("\n--- Challenge: Calculate Distance from User Input ---")

# Function to get a point's coordinates from user input
def get_point_from_user(point_name):
    while True:
        try:
            x_str = input(f"Enter x-coordinate for {point_name}: ")
            y_str = input(f"Enter y-coordinate for {point_name}: ")
            x = float(x_str) # Convert input to float to handle decimals
            y = float(y_str)
            return (x, y)
        except ValueError:
            print("Invalid input. Please enter numeric values for coordinates.")

# Get coordinates for point1 from user
point1_user = get_point_from_user("Point 1")

# Get coordinates for point2 from user
point2_user = get_point_from_user("Point 2")

print(f"User defined Point 1: {point1_user}")
print(f"User defined Point 2: {point2_user}")

# Calculate distance using the user-defined points
distance_user = math.sqrt((point2_user[0] - point1_user[0]) ** 2 + \
                           (point2_user[1] - point1_user[1]) ** 2)

print(f"Distance between user-defined point1 {point1_user} and point2 {point2_user}: {distance_user}")
```

## **Activity 4.1: List vs. Tuple: Scenario Matching**

การวิเคราะห์ว่าควรใช้ `list` หรือ `tuple` สำหรับแต่ละสถานการณ์ พร้อมเหตุผล:

1.  **ข้อมูลอุณหภูมิประจำวันที่มีการบันทึกไว้ตลอดหนึ่งสัปดาห์**
    *   **ทางเลือก:** `List`
    *   **เหตุผล:** อุณหภูมิสามารถเปลี่ยนแปลง อัปเดต หรืออาจมีการเพิ่มข้อมูลใหม่ได้ `list` สามารถเปลี่ยนแปลงได้ (mutable) ซึ่งอนุญาตให้มีการแก้ไขเหล่านี้ได้

2.  **ค่าสี RGB ของเฉดสีเฉพาะเฉดหนึ่ง (เช่น สีแดงคือ (255, 0, 0))**
    *   **ทางเลือก:** `Tuple`
    *   **เหตุผล:** ค่า RGB สำหรับสีเฉพาะโดยทั่วไปจะคงที่และไม่ควรเปลี่ยนแปลง `tuple` ไม่สามารถเปลี่ยนแปลงได้ (immutable) ซึ่งเหมาะสมสำหรับชุดข้อมูลที่คงที่

3.  **รายการสินค้าในรถเข็นชอปปิงที่สามารถกดเพิ่มหรือลบออกได้**
    *   **ทางเลือก:** `List`
    *   **เหตุผล:** สินค้าในรถเข็นมักจะถูกเพิ่ม ลบ หรือปริมาณอาจมีการอัปเดตบ่อยครั้ง `list` สามารถเปลี่ยนแปลงได้ (mutable) และรองรับการดำเนินการที่เปลี่ยนแปลงได้เหล่านี้

4.  **รายชื่อวันในหนึ่งสัปดาห์เรียงตามลำดับ (จันทร์ - อาทิตย์)**
    *   **ทางเลือก:** `Tuple`
    *   **เหตุผล:** ลำดับและเนื้อหาของวันในสัปดาห์มาตรฐานนั้นคงที่ `tuple` ช่วยให้ลำดับนี้คงที่และไม่สามารถเปลี่ยนแปลงได้โดยไม่ตั้งใจ

5.  **คะแนนเก็บของนักเรียนจากการส่งงานชุดต่างๆ ซึ่งอาจมีการอัปเดตหรือแก้ไขคะแนนในภายหลังได้**
    *   **ทางเลือก:** `List`
    *   **เหตุผล:** คะแนนของนักเรียนอาจมีการอัปเดต เพิ่มเติม (งานใหม่) หรือแก้ไข `list` สามารถเปลี่ยนแปลงได้ (mutable) และเหมาะสำหรับข้อมูลที่ต้องมีการเปลี่ยนแปลงเมื่อเวลาผ่านไป

6.  **พิกัด GPS ของสถานที่สำคัญหรือจุดแลนด์มาร์กที่ตั้งอยู่กับที่**
    *   **ทางเลือก:** `Tuple`
    *   **เหตุผล:** พิกัด GPS ของตำแหน่งที่คงที่ไม่เปลี่ยนแปลง `tuple` ไม่สามารถเปลี่ยนแปลงได้ (immutable) และสมบูรณ์แบบสำหรับการแสดงข้อมูลที่คงที่และไม่เปลี่ยนแปลง เช่น พิกัดทางภูมิศาสตร์

## **Activity 4.2: Interactive To-Do List**

Activity 4.2: Interactive To-Do List (using List Operations)
สิ่งที่อยากให้ทำ : ให้นักเรียนพัฒนาโปรแกรมจัดการรายการสิ่งที่ต้องทำ (To-Do List) แบบข้อความ (Text-based) อย่างง่าย โดยประยุกต์ใช้ความรู้เรื่องการจัดการ List ที่ได้เรียนมา
สิ่งที่ต้องมี
1. สร้าง List ว่างเริ่มต้นไว้ตัวหนึ่ง ชื่อว่า todo_list
2. ใช้การวนลูป while เพื่อแสดงเมนูคำสั่งให้ผู้ใช้เลือก ดังนี้:
2.1 เพิ่มรายการงาน (Add a task)
2.2 ดูรายการงานทั้งหมด (View tasks)
2.3 ทำเครื่องหมายว่างานเสร็จสิ้นแล้ว (ลบงานออก โดยเลือกจากตำแหน่ง Index หรือตามชื่อคำสั่ง)
2.4 ออกจากโปรแกรม (Exit)
3. ใช้คำสั่ง input() เพื่อรับค่าเมนูที่ผู้ใช้เลือก
4. ใช้คำสั่งจัดการ List ต่างๆ (เช่น append(), remove() หรือ pop(), len()) ในการบริหารจัดการข้อมูลใน todo_list

```python
todo_list = [] # 1. Create an empty list for to-do items

while True:
    print("\n--- To-Do List Menu ---")
    print("1. Add a task") # 2.1 Add a task
    print("2. View tasks")  # 2.2 View all tasks
    print("3. Mark task as completed (Remove)") # 2.3 Mark task as completed (Remove)
    print("4. Exit") # 2.4 Exit program

    # 3. Use input() to get user's menu choice
    choice = input("Enter your choice (1-4): ")

    if choice == '1':
        task = input("Enter the task to add: ")
        todo_list.append(task) # Use append()
        print(f"'{task}' added to your to-do list.")
    elif choice == '2':
        if not todo_list:
            print("Your to-do list is empty!")
        else:
            print("\n--- Your To-Do List ---")
            for i, task in enumerate(todo_list):
                print(f"{i + 1}. {task}") # Display tasks with numbers
    elif choice == '3':
        if not todo_list:
            print("Your to-do list is empty. Nothing to remove.")
            continue

        print("\n--- Tasks to remove ---")
        for i, task in enumerate(todo_list):
            print(f"{i + 1}. {task}")

        while True:
            try:
                remove_option = input("Enter the number of the task to mark as completed (or 'cancel'): ")
                if remove_option.lower() == 'cancel':
                    print("Task removal cancelled.")
                    break

                task_index = int(remove_option) - 1 # Adjust for 0-based indexing
                if 0 <= task_index < len(todo_list):
                    removed_task = todo_list.pop(task_index) # Use pop()
                    print(f"'{removed_task}' marked as completed and removed from the list.")
                    break
                else:
                    print("Invalid task number. Please try again.")
            except ValueError:
                print("Invalid input. Please enter a number or 'cancel'.")
    elif choice == '4':
        print("Exiting To-Do List program. Goodbye!")
        break
    else:
        print("Invalid choice. Please enter a number between 1 and 4.")
```
