# **Lab 3.1: โปรแกรมสร้างสูตรคูณ (Multiplication Table Generator)**

## Part 1: ตารางสูตรคูณแม่เดียว (Single Number Table):

````python
num_str = input("กรุณากรอกตัวเลขที่ต้องการสร้างตารางสูตรคูณ: ")
num = int(num_str)

for i in range(1, 13):
    print(f"{num} x {i} = {num * i}")
````

## Part 2: ตารางสูตรคูณแบบเต็ม (Full Multiplication Grid) - โจทย์ท้าทาย/ส่วนขยาย:

````python
for i in range(1, 13):  # ลูปนอก: กำหนดตัวเลขที่จะนำไปคูณ (ตั้งแต่ 1 ถึง 12)
    for j in range(1, 13):  # ลูปใน: กำหนดตัวคูณ (ตั้งแต่ 1 ถึง 12)
        print(f"{i * j:4d}", end="") # คำนวณผลคูณ i * j และจัดรูปแบบให้เว้น 4 ช่อง (ไม่ขึ้นบรรทัดใหม่)
    print() # เมื่อจบลูปใน (จบหนึ่งแถว) ให้ขึ้นบรรทัดใหม่
````

# **Lab 3.2: โปรแกรมการนับถอยหลัง & เกมทายตัวเลข (Countdown & User Guessing Game)**

## Part 1: โปรแกรมตัวนับถอยหลัง (Countdown Timer):

````python
start_num_str = input("กรุณากรอกตัวเลขเริ่มต้นสำหรับการนับถอยหลัง: ")
start_num = int(start_num_str)

current_num = start_num
while current_num >= 0:
    print(current_num)
    current_num -= 1

print("Blast off!")
````

## Part 2: เกมทายตัวเลขอย่างง่าย (Simple Guessing Game):

````python
import random

secret_number = random.randint(1, 100) # กำหนดตัวเลขลับระหว่าง 1 ถึง 100 (เช่น 42)

print("ฉันกำลังคิดตัวเลขระหว่าง 1 ถึง 100 ลองทายดูสิ!")

while True:
    try:
        guess_str = input("ทายตัวเลข: ")
        guess = int(guess_str)

        if guess < secret_number:
            print(f"{guess} น้อยเกินไป! ลองอีกครั้ง.")
        elif guess > secret_number:
            print(f"{guess} มากเกินไป! ลองอีกครั้ง.")
        else:
            print("ขอแสดงความยินดี! คุณทายถูกแล้ว!")
            break # ออกจากลูปเมื่อทายถูก

    except ValueError:
        print("โปรดกรอกตัวเลขที่ถูกต้อง")
````

### **โจทย์ท้าทาย (Challenge): เพิ่มการจำกัดจำนวนครั้งในการทาย**

````python
import random

secret_number = random.randint(1, 100) # กำหนดตัวเลขลับ

# ให้ผู้ใช้กำหนดจำนวนครั้งที่ทายได้เอง
while True:
    try:
        max_guesses_str = input("กรุณากำหนดจำนวนครั้งสูงสุดในการทาย: ")
        max_guesses = int(max_guesses_str)
        if max_guesses <= 0:
            print("จำนวนครั้งที่ทายต้องเป็นจำนวนเต็มบวก")
        else:
            break
    except ValueError:
        print("โปรดกรอกตัวเลขที่ถูกต้องสำหรับจำนวนครั้งสูงสุดในการทาย")

guesses_taken = 0

print(f"ฉันกำลังคิดตัวเลขระหว่าง 1 ถึง 100 คุณมีโอกาสทาย {max_guesses} ครั้ง")

while guesses_taken < max_guesses: # ลูปจะทำงานตราบเท่าที่จำนวนครั้งที่ทายยังไม่ถึงค่าสูงสุด
    try:
        guess_str = input(f"ทายตัวเลข (เหลือ {max_guesses - guesses_taken} ครั้ง): ")
        guess = int(guess_str)
        guesses_taken += 1

        if guess < secret_number:
            print(f"{guess} น้อยเกินไป! ลองอีกครั้ง.")
        elif guess > secret_number:
            print(f"{guess} มากเกินไป! ลองอีกครั้ง.")
        else:
            print("ขอแสดงความยินดี! คุณทายถูกแล้ว!")
            break # ออกจากลูปเมื่อทายถูก

    except ValueError:
        print("โปรดกรอกตัวเลขที่ถูกต้อง")

# ตรวจสอบว่าผู้ใช้ทายถูกหรือไม่หลังจากลูปสิ้นสุด
if guess != secret_number and guesses_taken == max_guesses:
    print("คุณทายครบจำนวนครั้งที่กำหนดแล้ว")
    print(f"ตัวเลขลับคือ {secret_number}")
````

## **Activity 3.1: Debugging Infinite Loops & Loop Logic Errors**

### **Buggy Snippet A: Infinite Loop**

**ข้อผิดพลาด:** ลูป `while` นี้เป็นลูปอนันต์ (Infinite Loop) เนื่องจากตัวแปรควบคุมลูป (`counter`) ไม่ได้ถูกอัปเดตค่าภายในลูป ทำให้เงื่อนไข `counter < 5` เป็นจริงเสมอและลูปจะทำงานไปเรื่อยๆ ไม่มีที่สิ้นสุด

**โค้ดที่มีข้อผิดพลาด (จะไม่รันเพื่อหลีกเลี่ยงลูปอนันต์):**
```python
counter = 0
while counter < 5:
    print(counter)
    # Missing counter increment!
```

**วิธีแก้ไข:** เพิ่มบรรทัด `counter += 1` ภายในลูป `while` เพื่อเพิ่มค่า `counter` ในแต่ละรอบ ทำให้เงื่อนไข `counter < 5` เป็นเท็จเมื่อ `counter` มีค่าถึง 5 และลูปจะสิ้นสุดลง

````python
# โค้ดที่แก้ไขแล้ว A: ลูปที่ไม่สิ้นสุด
counter = 0
while counter < 5:
    print(counter)
    counter += 1 # แก้ไข: เพิ่มค่า counter ในแต่ละรอบ

print("ลูปสิ้นสุดแล้ว")
````

### **Buggy Snippet B: Incorrect Range**

**ข้อผิดพลาด:** ฟังก์ชัน `range(start, stop)` ใน Python จะสร้างลำดับตัวเลขตั้งแต่ `start` ไปจนถึง `stop-1` (ไม่รวม `stop`) ดังนั้น `range(1, 5)` จะให้ค่า 1, 2, 3, 4 เท่านั้น แต่เป้าหมายคือต้องการพิมพ์ 1, 2, 3, 4, 5

**โค้ดที่มีข้อผิดพลาด:**

````python
# โค้ดที่มีข้อผิดพลาด B: ช่วงไม่ถูกต้อง
print("--- ผลลัพธ์จากโค้ดที่มีข้อผิดพลาด B ---")
for i in range(1, 5): # เป้าหมาย: พิมพ์ 1, 2, 3, 4, 5
    print(i)
````

**วิธีแก้ไข:** เพื่อให้ `range()` สร้างลำดับตัวเลขรวม 5 ด้วย เราต้องกำหนด `stop` เป็น 6 ซึ่งก็คือ `range(1, 6)`

````python
# โค้ดที่แก้ไขแล้ว B: ช่วงที่ถูกต้อง
print("--- ผลลัพธ์จากโค้ดที่แก้ไขแล้ว B ---")
for i in range(1, 6): # แก้ไข: เปลี่ยนเป็น range(1, 6) เพื่อรวมเลข 5 ด้วย
    print(i)
````

### **Buggy Snippet C: Misplaced Break**

**ข้อผิดพลาด:** โค้ดนี้มีเจตนาให้หยุดลูปเมื่อเจอตัวอักษร 'l' และแจ้งว่าพบ 'l' แต่เนื่องจากตำแหน่งของ `print(char)` อยู่ก่อนเงื่อนไข `if char == 'l'` ทำให้ตัวอักษร 'l' ถูกพิมพ์ออกมาก่อนที่จะมีการตรวจสอบและหยุดลูป ซึ่งอาจไม่ใช่พฤติกรรมที่ต้องการหากต้องการหยุดทันทีที่พบ 'l' โดยไม่พิมพ์ตัว 'l' นั้นออกมาก่อน

````python
# โค้ดที่มีข้อผิดพลาด C: การวาง break ผิดที่ (ตามคำอธิบาย)
print("--- ผลลัพธ์จากโค้ดที่มีข้อผิดพลาด C ---")
for char in "hello":
    print(char) # พิมพ์ตัวอักษรก่อนที่จะตรวจสอบเงื่อนไข break
    if char == 'l':
        print("พบ 'l', หยุด!")
        break
````

**วิธีแก้ไข:** ย้าย `print(char)` ไปอยู่หลังเงื่อนไข `if` และก่อน `break` เพื่อให้ `print(char)` ทำงานเฉพาะในกรณีที่ `char` ไม่ใช่ 'l' และลูปยังไม่ถูกหยุด

````python
# โค้ดที่แก้ไขแล้ว C: การวาง break ที่ถูกต้อง
print("--- ผลลัพธ์จากโค้ดที่แก้ไขแล้ว C ---")
for char in "hello":
    if char == 'l':
        print("พบ 'l', หยุด!")
        break
    print(char) # แก้ไข: พิมพ์ตัวอักษรก็ต่อเมื่อยังไม่เจอ 'l' และยังไม่หยุดลูป
````

## **Activity 3.2: Pattern Printing with Nested Loops**

ใช้ลูป `for` แบบซ้อน (nested for loops) เพื่อสร้างรูปแบบ ASCII art

---

### **รูปแบบ A: สี่เหลี่ยมของดอกจัน (A square of asterisks)**

เป้าหมายคือการพิมพ์สี่เหลี่ยมขนาด 5x5 โดยใช้เครื่องหมายดอกจัน (`*`)

````python
# โค้ดสำหรับพิมพ์สี่เหลี่ยมของดอกจัน
print("--- รูปแบบสี่เหลี่ยม ---")
for i in range(5):  # ลูปนอก: ควบคุมจำนวนแถว (5 แถว)
    for j in range(5):  # ลูปใน: ควบคุมจำนวนคอลัมน์ (5 ดอกจันในแต่ละแถว)
        print("*", end="") # พิมพ์ดอกจันโดยไม่ขึ้นบรรทัดใหม่
    print() # ขึ้นบรรทัดใหม่เมื่อจบลูปใน (จบหนึ่งแถว)
````

---

### **รูปแบบ B: สามเหลี่ยมมุมฉากของดอกจัน (A right-angled triangle of asterisks)**

เป้าหมายคือการพิมพ์สามเหลี่ยมมุมฉาก โดยจำนวนดอกจันจะเพิ่มขึ้นในแต่ละแถว

````python
# โค้ดสำหรับพิมพ์สามเหลี่ยมมุมฉากของดอกจัน
print("--- รูปแบบสามเหลี่ยมมุมฉาก ---")
for i in range(1, 6):  # ลูปนอก: ควบคุมจำนวนแถว (1 ถึง 5)
    for j in range(i):  # ลูปใน: ควบคุมจำนวนดอกจันในแต่ละแถว (เท่ากับค่า i ของแถวนั้นๆ)
        print("*", end="")
    print() # ขึ้นบรรทัดใหม่เมื่อจบลูปใน
````

### **โจทย์ท้าทาย (Challenge): สามเหลี่ยมมุมฉากกลับหัว (Inverted Right-Angled Triangle)**

เป้าหมายคือการพิมพ์สามเหลี่ยมมุมฉากที่กลับหัวลง โดยจำนวนดอกจันจะลดลงในแต่ละแถว

````python
# โค้ดสำหรับพิมพ์สามเหลี่ยมมุมฉากกลับหัว
print("--- รูปแบบสามเหลี่ยมมุมฉากกลับหัว ---")
for i in range(5, 0, -1):  # ลูปนอก: ควบคุมจำนวนแถว (จาก 5 ลงไป 1)
    for j in range(i):  # ลูปใน: ควบคุมจำนวนดอกจันในแต่ละแถว (เท่ากับค่า i ของแถวนั้นๆ)
        print("*", end="")
    print() # ขึ้นบรรทัดใหม่เมื่อจบลูปใน
````

### **โจทย์ท้าทาย (Challenge): ปิรามิด (Pyramid)**

เป้าหมายคือการพิมพ์รูปปิรามิดที่จัดกึ่งกลาง

````python
# โค้ดสำหรับพิมพ์รูปปิรามิด
print("--- รูปแบบปิรามิด ---")
num_rows = 5 # กำหนดจำนวนแถวของปิรามิด
for i in range(1, num_rows + 1):  # ลูปนอก: ควบคุมจำนวนแถว (จาก 1 ถึง num_rows)
    # พิมพ์ช่องว่างก่อนดอกจันเพื่อให้ปิรามิดอยู่ตรงกลาง
    print(" " * (num_rows - i), end="")
    # พิมพ์ดอกจัน (จำนวนดอกจันคือ 2*i - 1)
    print("*" * (2 * i - 1))
````
