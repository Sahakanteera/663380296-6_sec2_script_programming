# Lab 6.1: Simple Calculator Functions

```python
def add(a, b):
    """บวกตัวเลขสองจำนวนและคืนค่าผลลัพธ์"""
    return a + b

def subtract(a, b):
    """ลบตัวเลขสองจำนวน (a - b) และคืนค่าผลลัพธ์"""
    return a - b

def multiply(a, b):
    """คูณตัวเลขสองจำนวนและคืนค่าผลลัพธ์"""
    return a * b

def divide(a, b):
    """หารตัวเลขสองจำนวน (a / b) และจัดการกรณีหารด้วยศูนย์"""
    if b == 0:
        return "Error: Division by zero"
    return a / b

def power(base, exponent=2):
    """ยกกำลังตัวเลข โดยมีค่าเริ่มต้นของเลขชี้กำลังเป็น 2 (ยกกำลังสอง)"""
    return base ** exponent

# เมนูโปรแกรม (Menu-driven program)
while True:
    print("\n--- Simple Calculator ---")
    print("1. Add (+)")
    print("2. Subtract (-)")
    print("3. Multiply (*)")
    print("4. Divide (/)")
    print("5. Power (^)")
    print("6. Exit")

    choice = input("Select an operation (1-6): ")

    if choice == '6':
        print("Exiting program. Goodbye!")
        break

    if choice in ['1', '2', '3', '4', '5']:
        try:
            num1 = float(input("Enter first number (or base): "))

            if choice == '5':
                #Challenge: ฟังก์ชัน Power
                exp_input = input("Enter exponent (press Enter to use default '2'): ")
                if exp_input == "":
                    result = power(num1)
                else:
                    num2 = float(exp_input)
                    result = power(num1, num2)
            else:
                #คำนวณปกติ 1-4
                num2 = float(input("Enter second number: "))

                if choice == '1':
                    result = add(num1, num2)
                elif choice == '2':
                    result = subtract(num1, num2)
                elif choice == '3':
                    result = multiply(num1, num2)
                elif choice == '4':
                    result = divide(num1, num2)

            print(f"Result: {result}")

        except ValueError:
            print("Invalid input! Please enter numeric values.")
    else:
        print("Invalid choice. Please select from 1 to 6.")
```

## my_utils

```python
%%writefile my_utils.py

def greet(name):
    """พิมพ์ข้อความทักทาย"""
    print(f"Hello, {name}!")

def is_prime(number):
    """ตรวจสอบว่าเป็นจำนวนเฉพาะหรือไม่ (คืนค่า True หากเป็น จำนวนเฉพาะ)"""
    if number <= 1:
        return False
    # ตรวจสอบการหารลงตัวตั้งแต่ 2 ถึง รากที่สองของตัวเลข
    for i in range(2, int(number**0.5) + 1):
        if number % i == 0:
            return False
    return True
```

### ตัวอย่างการใช้งาน 3 Standard Library ในโลกจริง

*   **1. โมดูล `datetime`:**
    *   **การทำงาน:** ใช้ดึงข้อมูลวัน/เวลาปัจจุบัน และจัดรูปแบบ (Format) ให้อ่านง่าย
    *   **การนำไปใช้งานจริง:** ใช้ประทับเวลา (Timestamp) ว่าข้อมูลถูกบันทึกเมื่อไหร่ เช่น เวลาที่ผู้ใช้ทำรายการสั่งซื้อสำเร็จ หรือเวลาที่เกิดข้อผิดพลาดในระบบ (Log)

*   **2. โมดูล `os`:**
    *   **การทำงาน:** ใช้ระบุตำแหน่งที่อยู่ปัจจุบัน (Current Working Directory) ว่าโปรแกรมรันอยู่โฟลเดอร์ใด
    *   **การนำไปใช้งานจริง:** จำเป็นมากในการเขียนโปรแกรมที่ต้อง อ่าน/เขียนไฟล์ เพื่อป้องกันปัญหาโปรแกรมหาตำแหน่งไฟล์รูปภาพหรือไฟล์ CSV ไม่เจอ

*   **3. โมดูล `collections` (คำสั่ง `Counter`):**
    *   **การทำงาน:** ใช้นับความถี่ของข้อมูลในชุดข้อมูลแบบอัตโนมัติ
    *   **การนำไปใช้งานจริง:** ช่วยลดการเขียนลูปยาว ๆ มักใช้ในการวิเคราะห์ข้อมูล (Data Analysis) เช่น การหาคำที่พบบ่อยที่สุดในบทความ หรือระบบนับคะแนนโหวต

```python
# ตัวอย่างการใช้งาน 3 Standard Library ที่น่าสนใจ
import datetime
import os
from collections import Counter

# ---------------------------------------------------------
# 1. โมดูล datetime: สำหรับจัดการวันที่และเวลา
# ---------------------------------------------------------
print("--- 1. datetime Module ---")
# ดึงวันและเวลาปัจจุบัน
now = datetime.datetime.now()
# จัดรูปแบบการแสดงผลให้อ่านง่าย
formatted_time = now.strftime("%d/%m/%Y %H:%M:%S")
print(f"วันและเวลาปัจจุบัน: {formatted_time}")


# ---------------------------------------------------------
# 2. โมดูล os: สำหรับจัดการระบบไฟล์และระบบปฏิบัติการ
# ---------------------------------------------------------
print("\n--- 2. os Module ---")
# ดูว่าตอนนี้สคริปต์กำลังรันอยู่ที่โฟลเดอร์ไหน (Current Working Directory)
current_dir = os.getcwd()
print(f"ตำแหน่งโฟลเดอร์ปัจจุบัน:\n{current_dir}")


# ---------------------------------------------------------
# 3. โมดูล collections: โครงสร้างข้อมูลขั้นสูง
# ---------------------------------------------------------
print("\n--- 3. collections Module ---")
text = "programming"
# ใช้ Counter เพื่อนับจำนวนตัวอักษรแต่ละตัวในข้อความแบบอัตโนมัติ
letter_counts = Counter(text)

print(f"ผลการนับตัวอักษรในคำว่า '{text}':")
# แสดงผลลัพธ์การนับ
for letter, count in letter_counts.items():
    print(f"ตัวอักษร '{letter}' มี {count} ตัว")
```
